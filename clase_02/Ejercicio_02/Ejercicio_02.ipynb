1 — Cargar el dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
autos = pd.read_csv("Automobile.csv")
```

```python
autos.head()
```

2 — Gráfico de dispersión.

```python
plt.scatter(autos["weight"], autos["mpg"])

plt.xlabel("Peso del automóvil")
plt.ylabel("Consumo de combustible (mpg)")
plt.title("Relación entre peso y consumo")

plt.grid()
plt.show()
```

3 — Histograma.

```python
plt.hist(autos["mpg"])

plt.xlabel("Consumo de combustible (mpg)")
plt.ylabel("Cantidad de automóviles")
plt.title("Distribución del consumo de combustible")

plt.grid()
plt.show()
```

```python
autos["origin"] = autos["origin"].map({
    "usa": 1,
    "europe": 2,
    "japan": 3
})
```

4 — Boxplot.

```python
plt.boxplot([autos["cylinders"], autos["origin"]])

plt.xticks([1, 2], ["Cilindros", "Origen"])
plt.ylabel("Valor")
plt.title("Valores atípicos en cilindros y origen")

plt.grid()
plt.show()
```

5 — Distribución por año y origen.

```python
conteo = autos.groupby(["model_year", "origin"]).size().unstack(fill_value=0)

conteo.plot(kind="bar")

plt.xlabel("Año del modelo")
plt.ylabel("Cantidad de automóviles")
plt.title("Automóviles por año de modelo y origen")

plt.legend(title="Origen")
plt.grid()
plt.show()
```
