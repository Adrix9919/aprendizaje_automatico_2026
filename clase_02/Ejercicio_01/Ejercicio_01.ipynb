```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
ventas = pd.read_csv("ventas.csv")

print(ventas.head())
```

```python
clientes = pd.read_json("clientes.json")

print("Cantidad total de clientes:", len(clientes))
```

```python
inventario = pd.read_excel("inventario.xlsx")

promedio_stock = inventario["stock_disponible"].mean()

print("Promedio de stock:", promedio_stock)
```

```python
plt.plot(ventas["total_venta"], label="Total de venta")

plt.xlabel("Venta")
plt.ylabel("Monto")
plt.title("Ventas realizadas")

plt.legend()
plt.grid()

plt.show()
```

```python
plt.scatter(ventas["cantidad"], ventas["total_venta"])

plt.xlabel("Cantidad vendida")
plt.ylabel("Total de venta")
plt.title("Relación entre cantidad y total de venta")

plt.grid()
plt.show()
```
